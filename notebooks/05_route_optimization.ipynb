## Smart Waste Management & Collection Optimization

Implements FR-5: Haversine distance matrix, Nearest Neighbor heuristic, Google OR-Tools TSP solver, and benchmark comparisons.


```python
import sys
sys.path.append('..')
import pandas as pd
from src.route_optimizer import DEFAULT_DEPOT, solve_nearest_neighbor_tsp, solve_ortools_tsp, compare_routes
```

### 1. Select Priority Bins for Collection


```python
df = pd.read_csv('../data/processed/bin_readings_cleaned.csv')
latest_df = df.drop_duplicates(subset=['bin_id'], keep='last')
# Filter critical & high bins
priority_bins = latest_df[latest_df['fill_level'] >= 65.0].to_dict(orient='records')
print(f'Total priority stops: {len(priority_bins)}')
```

### 2. Solve TSP using Nearest Neighbor Heuristic


```python
nn_res = solve_nearest_neighbor_tsp(DEFAULT_DEPOT, priority_bins, return_to_depot=True)
print(f'Nearest Neighbor Distance: {nn_res["total_distance_km"]} km')
```

### 3. Solve TSP using Google OR-Tools


```python
ortools_res = solve_ortools_tsp(DEFAULT_DEPOT, priority_bins, return_to_depot=True)
print(f'OR-Tools Distance: {ortools_res["total_distance_km"]} km')
```

### 4. Benchmark & Distance Reduction Metrics (FR-5.4)


```python
comparison = compare_routes(DEFAULT_DEPOT, priority_bins, ortools_res)
for metric, val in comparison.items():
    print(f'{metric}: {val}')
```
