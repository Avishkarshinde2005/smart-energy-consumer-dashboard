```python
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.analysis import compute_summary_kpis, compute_zone_statistics, compute_waste_type_breakdown
```

### 1. Load Cleaned Dataset & Compute High-Level KPIs


```python
df = pd.read_csv('../data/processed/bin_readings_cleaned.csv')
latest_df = df.drop_duplicates(subset=['bin_id'], keep='last')
kpis = compute_summary_kpis(latest_df)
for k, v in kpis.items():
    print(f'{k}: {v}')
```

### 2. Fill-Level Distribution Histogram


```python
plt.figure(figsize=(8, 4))
sns.histplot(latest_df['fill_level'], bins=15, kde=True, color='forestgreen')
plt.axvline(85, color='red', linestyle='--', label='Critical Threshold (86%)')
plt.title('Distribution of Current Bin Fill Levels')
plt.xlabel('Fill Level (%)')
plt.ylabel('Count of Bins')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 3. Waste Type Composition


```python
waste_counts = latest_df['waste_type'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(waste_counts, labels=waste_counts.index, autopct='%1.1f%%', colors=['#4caf50', '#2196f3', '#ff9800', '#e91e63'])
plt.title('Fleet Waste Stream Breakdown')
plt.show()
```
